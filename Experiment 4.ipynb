```python
cnt=0
s="My Name is Anuja Sanjay Patil"
vowels = "aeiouAEIOU"
c=set()
for char in s:
    if char in vowels and char not in c:
      cnt += 1

print("count of vowels:",cnt)
```

```python
cnt=0
s="My Name is Anuja Sanjay Patil"
vowels = "aeiouAEIOU"
counted_chars = set()
for char in s:
    if char in vowels and char not in counted_chars:
      cnt += 1
      counted_chars.add(char)
print("count of unique vowels:",cnt)
```

```python
s = "My Name is Anuja Sanjay Patil"
char_counts = {}

for char in s:
    if char in char_counts:
        char_counts[char] += 1
    else:
        char_counts[char] = 1

print("Character counts:", char_counts)
```
